```python
import pandas as pd
import re

# Load datasets
file_path1 = "/content/sample_data/Student depression and anxiety dataset.xlsx"
file_path2 = "/content/sample_data/depression_dataset_reddit_cleaned.csv"
file_path3 = "/content/sample_data/Stress.csv"
file_path4 = "/content/sample_data/Suicide_Ideation_Dataset.csv"

df1 = pd.read_excel(file_path1)
df2 = pd.read_csv(file_path2)
df3 = pd.read_csv(file_path3)
df4 = pd.read_csv(file_path4, encoding='latin1') # Added encoding='latin1'


# Function to clean text
def clean_text(text):
    text = str(text).lower()  # Convert to lowercase
    text = re.sub(r'http\S+|www\S+|https\S+', '', text)  # Remove URLs
    text = re.sub(r'[^a-z\s]', '', text)  # Remove punctuation, numbers, special chars
    text = re.sub(r'\s+', ' ', text).strip()  # Remove extra spaces
    return text

# Function to clean dataset safely
def clean_dataset(df, text_col='text', label_col='label'):
    # IMPORTANT: Make a copy to avoid modifying the original dataframe passed to the function
    df_temp = df.copy()

    # Determine the actual text column name from the dataframe
    actual_text_col = text_col if text_col in df_temp.columns else df_temp.columns[0]

    # Check if the specified label column exists
    actual_label_col = label_col
    if label_col and label_col not in df_temp.columns:
        print(f"Warning: '{label_col}' column not found in dataset. Skipping label processing for this dataset.")
        actual_label_col = None # Set to None if label_col does not exist

    # Apply text cleaning
    df_temp['cleaned_text'] = df_temp[actual_text_col].apply(clean_text)

    # Select columns for the cleaned output
    if actual_label_col:
        # Print head with original text, cleaned text, and label
        print(df_temp[[actual_text_col, 'cleaned_text', actual_label_col]].head())
        # Return only the label and cleaned text
        return df_temp[[actual_label_col, 'cleaned_text']]
    else:
        # Print head with original text and cleaned text
        print(df_temp[[actual_text_col, 'cleaned_text']].head())
        # Return only the cleaned text
        return df_temp[['cleaned_text']]

# Apply cleaning to all datasets
print("Cleaning Dataset 1:")
df1_cleaned = clean_dataset(df1)

print("\n Cleaning Dataset 2:")
df2_cleaned = clean_dataset(df2, text_col='clean_text', label_col='is_depression')

print("\n Cleaning Dataset 3:")
df3_cleaned = clean_dataset(df3, text_col='text', label_col='label')

print("\n Cleaning Dataset 4:")
df4_cleaned = clean_dataset(df4, text_col='Tweet', label_col='Suicide')
# Convert 'Suicide' column values for df4_cleaned: 'Not Suicide post' = 0 and else = 1
df4_cleaned['Suicide'] = df4_cleaned['Suicide'].apply(lambda x: 0 if x == 'Not Suicide post' else 1)


# Save cleaned versions
df1_cleaned.to_csv("/content/dataset1.csv", index=False)
df2_cleaned.to_csv("/content/dataset2.csv", index=False)
df3_cleaned.to_csv("/content/dataset3.csv", index=False)
df4_cleaned.to_csv("/content/dataset4.csv", index=False)

print("\nCleaned datasets saved successfully!")
```
